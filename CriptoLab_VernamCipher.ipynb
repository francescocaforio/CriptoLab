Il cifrario di Vernam è stato inventato nel 1917 da Gilbert Vernam.   Funziona così:

- il **testo in chiaro** viene convertito in binario (sequenza di 0 e 1);
- si genera una **chiave segreta**, lunga quanto il messaggio;
- si applica l’operazione logica **XOR** tra ogni bit del messaggio e della chiave;
- il risultato è il **testo cifrato**;
- per **decifrare**, basta applicare di nuovo XOR tra il testo cifrato e la stessa chiave.

Se la chiave è:
- lunga quanto il messaggio,
- usata una sola volta,
- completamente casuale,

allora otteniamo il **One-Time Pad (OTP)**, che è **teoricamente inviolabile**.


**Importante**
* La funzione `string_to_binary(s)` prende una stringa e la converte in **binario ASCII a 8 bit**.  In questo modo possiamo rappresentare qualsiasi testo come una sequenza di zeri e uni.
* la funzione `binary_to_string(b)` fa l’operazione inversa: prende una lunga sequenza binaria, la divide in gruppi da 8 bit, e converte ogni gruppo nel corrispondente carattere ASCII. Così possiamo tornare dal messaggio cifrato (in binario) al testo leggibile.
* La funzione `genera_chiave_binaria(lunghezza)` genera una chiave **casuale** composta da 0 e 1. La chiave deve avere la **stessa lunghezza del messaggio in binario**.  Questo è fondamentale per applicare lo XOR bit a bit.
* La funzione `cifra_vernam(testo)` converte il testo in binario (`string_to_binary`), genera una chiave casuale e applica **XOR bit a bit** tra testo e chiave. Restituisce: il testo cifrato in binario e la chiave usata.
* La funzione `decifra_vernam(cifrato, chiave)` è **identica alla cifratura**: si applica di nuovo lo XOR tra il testo cifrato e la chiave e si ottiene il messaggio originale in binario che poi viene riconvertito in caratteri.


```python
import random

# Funzione per convertire una stringa in binario ASCII a 8 bit
def string_to_binary(s):
    return ''.join(format(ord(c), '08b') for c in s)

# Funzione per convertire una stringa binaria a caratteri
def binary_to_string(b):
    return ''.join(chr(int(b[i:i+8], 2)) for i in range(0, len(b), 8))

# Genera una chiave casuale lunga quanto il messaggio, in formato binario
def genera_chiave_binaria(lunghezza):
    return ''.join(random.choice('01') for _ in range(lunghezza))

# Funzione di cifratura XOR bit a bit
def cifra_vernam(testo):
    binario_testo = string_to_binary(testo)
    print(binario_testo)
    chiave = genera_chiave_binaria(len(binario_testo))
    print(chiave)
    cifrato = ''.join(str(int(b) ^ int(k)) for b, k in zip(binario_testo, chiave))
    return cifrato, chiave

# Funzione di decifratura XOR (identica a cifratura in Vernam)
def decifra_vernam(cifrato, chiave):
    binario_testo = ''.join(str(int(c) ^ int(k)) for c, k in zip(cifrato, chiave))
    return binary_to_string(binario_testo)
```

```python
# Esempio d'uso

messaggio = "Cifrario di Vigenere"
cifrato, chiave = cifra_vernam(messaggio)
decifrato = decifra_vernam(cifrato, chiave)

print("Messaggio originale:", messaggio)
print("Chiave binaria:      ", chiave)
print("Testo cifrato:       ", cifrato)
print("Testo decifrato:     ", decifrato)
```
